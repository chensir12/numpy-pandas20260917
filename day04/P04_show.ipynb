# 正余弦曲线--方式1 MATLAB风格

```python
from unittest import case

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

#准备数据
x = np.linspace(0,10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

#画图
plt.figure(figsize=(10,6))
plt.subplot(2,1,1)
plt.xlim(0,10)
plt.ylim(-1,1)
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('sin')
plt.plot(x,y1)

plt.subplot(2,1,2)
plt.xlim(0,10)
plt.ylim(-1,1)
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.title('cos(x)')
plt.plot(x,y2)
```

# 正余弦曲线 --方式2   面向对象的方式

```python
# 准备数据
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots(2,figsize=(10,6))
# 对画布第一个区域进行绘图
ax[0].set_xlim(0,10)
ax[0].set_ylim(-1,1)
ax[0].set_xlabel('x')
ax[0].set_ylabel('sin(x)')
ax[0].set_title('sin')
ax[0].plot(x,y1)

ax[1].set_xlim(0,10)
ax[1].set_ylim(-1,1)
ax[1].set_xlabel('x')
ax[1].set_ylabel('cos(x)')
ax[1].set_title('cos')
ax[1].plot(x,y2)
```

```python
#多变量可视化
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams

rcParams["font.sans-serif"] = ["PingFang SC", "Heiti SC", "Arial Unicode MS"]
rcParams["axes.unicode_minus"] = False

#处理series是一个元素一个元素处理
def year_color(x):
    """添加一列，为不同年份的数据添加不同的颜色"""
    match x.year:
        case 2012:
            return "r"
        case 2013:
            return "g"
        case 2014:
            return "b"
        case 2015:
            return "k"

df = pd.read_csv(r"/Users/blucechen/PycharmProjects/pythontest/data/weather.csv")
df["date"] = pd.to_datetime(df["date"])
df["color"] = df["date"].apply(year_color)
fig = plt.figure()
ax1 = fig.add_subplot(1,1,1)
#绘制散点图，横轴为最高气温，纵轴为降水量
#c设置颜色，alpha设置透明度
ax1.scatter(df["temp_max"], df["precipitation"],c=df["color"],alpha=0.5)
ax1.set_xlabel("最高气温")
ax1.set_ylabel("降水量")
```

# Pandas可视化

```python
import pandas as pd

df = pd.read_csv("/Users/blucechen/PycharmProjects/pythontest/data/sleep.csv")
df.info()
```

```python
# 使用柱状图展示不同睡眠时长的数量
pd.cut(df["sleep_duration"],[0,5,6,7,8,9,10,11,12]).value_counts().plot(kind="bar",color=["red", "green", "blue", "yellow", "cyan", "magenta", "black", "purple"])

pd.cut(df["sleep_duration"],[0,5,6,7,8,9,10,11,12]).value_counts().plot.bar(color=["red", "green", "blue", "yellow", "cyan", "magenta", "black", "purple"])

```

```python
# 使用折线图展示不同睡眠时长的数量
pd.cut(df["sleep_duration"],[0,5,6,7,8,9,10,11,12]).value_counts().sort_index().plot()
```

```python
# 使用面积图展示不同睡眠时长的数量
pd.cut(df["sleep_duration"],[0,5,6,7,8,9,10,11,12]).value_counts().sort_index().plot.area()
```

```python
# 使用直方图展示不同睡眠时长的数量
df["sleep_duration"].value_counts().plot.hist()
```

```python
# 使用饼状图展示不同睡眠时长的占比
pd.cut(df["sleep_duration"],[0,5,6,7,8,9,10,11,12]).value_counts().sort_index().plot.pie()
```

# 双变量可视化

```python
df.plot.scatter(x="sleep_duration",y="sleep_quality")
```

```python
df.plot(kind="scatter",x="sleep_duration",y="sleep_quality")
```

```python
#绘制睡眠时间与睡眠质量的蜂窝图
df.plot.hexbin(x="sleep_duration",y="sleep_quality",gridsize=10)
```

```python
# 堆叠图
df["sleep_quality_stage"] = pd.cut(df["sleep_quality"],range(11))
# df["sleep_duration_stage"]
df["sleep_duration_stage"] = pd.cut(df["sleep_duration"],[0,5,6,7,8,9,10,11,12])
df_pivot_table = df.pivot_table(
    values="person_id",index="sleep_quality_stage",columns="sleep_duration_stage",aggfunc="count",
    observed=True
)

df_pivot_table.plot.bar()

df_pivot_table.plot.bar(stacked=True)
```
