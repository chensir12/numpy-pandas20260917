# 导包 读取数据

```python
#导包
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams

rcParams["font.sans-serif"] = ["PingFang SC", "Heiti SC", "Arial Unicode MS"]
rcParams["axes.unicode_minus"] = False

# 从指定的数据源中读取房屋数据
data = pd.read_csv("/Users/blucechen/PycharmProjects/pythontest/data/house_sales.csv")
data
```

# 数据清洗

```python
# 检查缺失值
missing_values = data.isnull().sum()
print('各列缺失值数量:')
print(missing_values)
```

```python
#过滤掉缺失值
data = data.dropna()
data

# 检测异常值 使用IQR方法
Q1 = data["price"].quantile(0.25)
Q3 = data["price"].quantile(0.75)
IQR = Q3 - Q1
low_bound = Q1 - 1.5 * IQR
high_bound = Q3 + 1.5 * IQR

data = data[(data["price"] >= low_bound) & (data["price"] <= high_bound)]
data
```

# 数据类型转换 将日期字符串转换为日期对象

```python
data["date"] = pd.to_datetime(data["date"])
data.info()
```

# 添加新的特征 房龄、是否翻新

```python
# 添加房龄字段
data["age"] = data["date"].dt.year - data["yr_built"]

# 添加是否翻新
data["is_renovated"] = data["yr_renovated"].apply(lambda x: 1 if x > 0 else 0)
data
```

# 数值型列描述性统计

```python
numeric_columns = data.select_dtypes(include=[np.number]).columns

data[numeric_columns].describe()
```

# 数值型列的相关性统计

```python
data[numeric_columns].corr()
```

# 按邮政编码分组，计算每组的平均房价、平均居住面积、平均卧室数量

```python
data.groupby("zipcode")[["price","sqft_living","bedrooms"]].mean()
```

# 按是否翻新分组，计算每组的平均房价、最大居住面积、最小卧室数量

```python
data.groupby("is_renovated").agg({"price":"mean","sqft_living":"max","bedrooms":"min"})
```

# 按房屋使用年限分组(简单分为5个区间)

```python
data["age_group"] = pd.cut(data["age"],5)
data["age_group"]
data.pivot_table(values=["price","sqft_living","bedrooms"],index="age_group",aggfunc="mean",observed=False)
```

# 时间序列分析-每年平均房价

```python
data.groupby(data["date"].dt.year)["price"].mean()
```

# 时间序列分析-不同翻新情况平均房价

```python
data.pivot_table(values="price",index=[data["date"].dt.year,"is_renovated"],aggfunc="mean")
data.pivot_table(values="price",index=data["date"].dt.year,columns="is_renovated",aggfunc="mean")
```

# 可视化 房价分布直方图

```python
fig = plt.figure(figsize = (10,10))
plt.hist(data["price"],bins=30,edgecolor="k")
plt.title('房价分布直方图')
plt.xlabel('房价')
plt.ylabel('频数')
```
