### Series常用属性

```python
import pandas as pd
arrs = pd.Series([11,22,33,44,55],name="atguigu",index=['a','b','c','d','e'])
print(arrs.index) #索引对象
for i in arrs.index:
    print(i)
print(arrs.values) #Series的值
print(arrs.ndim) #Series的维度
print(arrs.shape) #Series的形状
print(arrs.size) #Series的元素个数
print(arrs.dtype) #Series的元素类型
print(arrs.dtypes) #Series的元素类型
print(arrs.name) #Series的名称
print(arrs.loc['c']) #显式索引，按标签索引切片
print(arrs.loc['c':'d']) #显式索引，按标签索引切片
print(arrs.iloc[0]) #隐式索引，按标签索引切片
print(arrs.iloc[0:3]) #隐式索引，按标签索引切片
print(arrs.at['a']) #使用标签访问单个元素
print(arrs.iat[3]) #使用位置访问单个元素
```

### Series 常用方法

```python
import pandas as pd
import numpy as np
#np.nan 这是一个float类型
arrs = pd.Series([11,22,np.nan,None,44,22],index=['a','b','c','d','e','f'])
arrs.head(2) #查看前n行数据，默认5行 基于原对象操作的
arrs.tail(2) #查看后N行数据,默认五行
arrs.isin([11,22]) #元素是否包含在参数列表集合中
arrs.isna() #元素知否为缺失值(NaN 或 None)
arrs.sum() # 求和，忽略缺失值
arrs.mean() #平均值
arrs.min() #最小值
arrs.max() #最大值
arrs.var() #方差
arrs.std() #标准差
arrs.median() #中位数:奇数个数取中间数 偶数个数中间两个数加起来除以2
arrs.mode() #众数(出现频率最高的值)
arrs.quantile(0.25,interpolation='linear') #分位数 插值方式可选 linear lower higher midpoint nearest
arrs.describe() #常见统计信息
arrs.value_counts() #每个元素的个数
arrs.count() #非缺失元素的个数
len(arrs) #包涵缺失元素的个数
type(arrs.drop_duplicates()) #去重后的series
type(arrs.unique()) #去重后的数组
arrs.nunique() #去重后元素的个数
arrs.sample() #随机采样
# arrs = pd.Series([11,22,np.nan,None,44,22],index=['f','a','c','d','e','b']) #按索引排序
arrs.sort_index()
arrs.sort_values()
arrs.replace(np.nan,66) #用指定值代替原有值
# print(type(arrs))
arrs.to_frame() #将Series 转换为DataFrame
# print(type(arrs.to_frame()))
arr1 = pd.Series([1,2,3])
arr2 = pd.Series([1,2,3])
arr1.equals(arr2) #判断两个Series是否相同
arrs.keys() #返回Series的索引对象 和index的效果一样
arrs.index
# corr() 计算与另一个Series的相关系数
# 由于 arr1 和 arr2 的值完全相同，它们之间是完全正相关的，
#因此相关系数为 1。
# arr1.corr(arr3)：arr1 的值是递增的，而 arr3 的值是递减的，它们之间是完全
# 负相关的，所以相关系数为 -1。
# arr1.corr(arr4)：arr1 和 arr4 的值都是递增的，且变化趋势一致，它们之间是
# 完全正相关的，相关系数为 1。
# arr5.corr(arr6)：arr5 和 arr6 的值之间没有明显的线性关系，它们的相关系数
# 为 0。
arr3 = pd.Series([3,2,1])
arr4 = pd.Series([6,7,8])
arr5 = pd.Series([1, -1, 1, -1])
arr6 = pd.Series([1, 1, -1, -1])
# print(arr1.corr(arr2))
# print(arr1.corr(arr3))
# print(arr1.corr(arr4))
print(arr5.corr(arr6))
# cov() 计算与另一个Series的协方差
# 协方差用于衡量两个变量的总体误差，其值的正负表示两个变量的变化方向关系：
# 正值表示同向变化，负值表示反向变化。
print(arr1.cov(arr3))
# hist() 绘制直方图
# 直方图（Histogram）是一种用于展示数据分布的统计图表，它通过将数据划分为
# 若干个连续的区间（ bins ），统计每个区间内数据的频数或频率，并用矩形条的
# 高度表示该区间的数值分布情况
arr7 = pd.Series([3,2,1,1,1,2,2])
# arr7.hist(bins=3)
# items()  获取索引名以及值
for i,v in arr7.items():
    print(i,v)
```

### Series的布尔索引

```python
import pandas as pd
s = pd.Series({"a":-1.2,"b":3.5,"c":6.8,"d":2.9})
print(s.mean())
bools = s > s.mean()
print(bools)
print(s[bools])
```

```python
import pandas as pd
df = pd.DataFrame(
    data={"id": [101, 102, 103], "name": ["张三", "李四", "王五"], "age": [20, 30, 40]},
    index=["aa", "bb", "cc"])
print(df.index)
print(df.columns)
print(df.values)
print(df.ndim)
print(df.shape)
print(df.size)
print(df.dtypes)
print(df.T)
# loc[] 显式索引，按行列标签索引或切片 逗号前是行切片规则，后是列切片规则
print(df.loc["aa":"bb"]) #前闭后闭
print(df.loc[:,["id","name"]])
# iloc[] 隐式索引，按行列位置索引或切片
print(df.iloc[0:2]) #隐式索引就是前闭后开
print(df.iloc[0:2,2])
print('---------------')
# at[]行列标签访问单个元素
print(df.at['aa',"name"])
# iat[]使用行列位置访问单个元素
print(df.iat[0,1])
```
